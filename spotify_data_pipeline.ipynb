```python
auth_manager = SpotifyOAuth(
    client_id=CLIENT_ID,
    client_secret=CLIENT_SECRET,
    redirect_uri=REDIRECT_URI,
    scope="playlist-read-private playlist-read-collaborative",
    open_browser=False
)

# This prints the URL you need to open
print("Open this URL in your browser:")
theurl = auth_manager.get_authorize_url()
print(theurl)
```

```python
redirected_url = input("Paste the full URL from your browser here: ")
code = auth_manager.parse_response_code(redirected_url)
token = auth_manager.get_access_token(code, as_dict=False)
token=token
sp = spotipy.Spotify(auth=token)
print("Authenticated!")
```

```python
import pandas as pd

rows = []

for obj in results["items"]:
    track = obj["item"]

    rows.append({
        "track_id": track["id"],
        "track_name": track["name"],
        "album_id": track["album"]["id"],
        "album_name": track["album"]["name"],
        "artist_ids": ", ".join([a["id"] for a in track["artists"]]),
        "artists": ", ".join([a["name"] for a in track["artists"]]),
        "duration_ms": track["duration_ms"],
        "release_date": track["album"]["release_date"],
        "spotify_url": track["external_urls"]["spotify"]
    })

df = pd.DataFrame(rows)
```
